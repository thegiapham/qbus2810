Use this notebook to create a group-specific 80/20 train/test split for the bike sharing group assignment.

- Replace `state` with the sum of the student ID numbers for your group members.
- Use `train` for exploratory analysis, estimation, diagnostics, and model selection.
- Use `test` only for the final forecast comparison.
- For predictive tasks, think of the forecast as being made before any user counts for that day are observed.
- The split is group-specific, but you still need to decide which variables are admissible for each modelling goal.


```python
import pandas as pd

data = pd.read_csv('bike_sharing_day_2026.csv')

state = 1108224  # replace with the sum of your group members' student IDs

train = data.sample(frac=0.8, random_state=state)
test = data.loc[~data.index.isin(train.index)].copy()

train = train.reset_index(drop=True)
test = test.reset_index(drop=True)

print(train.shape)
print(test.shape)
```

```python
# Optional: save group-specific copies
train.to_csv('train_group.csv', index=False)
test.to_csv('test_group.csv', index=False)
```
